# Data Preprocessing Tools

## Importing libraries

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.impute import SimpleImputer
```

## Imporint dataset

```python
dataset = pd.read_csv("../Datasets/Data.csv")
```

```python
x = dataset.iloc[:, :-1].values
y = dataset.iloc[:, -1].values
```

```python
print(x)
```

```python
print(y)
```

## Taking care Missing Data

```python
x[:,1:3]
```

```python
imputer = SimpleImputer(missing_values=np.nan,strategy="mean")
imputer.fit(x[:,1:3])
x[:, 1:3] = imputer.transform(x[:, 1:3])
```

```python
print(x)
```

## Encoding Categorical Data

### Encoding Independent Variable

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
```

```python
ct = ColumnTransformer(transformers=[('encoder', OneHotEncoder(), [0])], remainder='passthrough')
```

```python
x = np.array(ct.fit_transform(x))
```

```python
print(x)
```

### Enncoding Dependent Variable

```python
from sklearn.preprocessing import LabelEncoder
```

```python
le = LabelEncoder()
y = le.fit_transform(y)
```

```python
print(y)
```

## Splitting dataset into Training set and Test set

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
print("Original feature size:", len(x), "rows")
print("Training feature size:", len(X_train), "rows")
print("Testing feature size:", len(X_test), "rows")
```

```python
print("Original Target size:", len(y), "rows")
print("Training Target size:", len(y_train), "rows")
print("Testing Target size:", len(y_test), "rows")
```

## Feature Scaling

```python
from sklearn.preprocessing import StandardScaler
```

```python
sc = StandardScaler()
X_train[:,3:] = sc.fit_transform(X_train[:,3:])
X_test[:,3:] = sc.transform(X_test[:,3:])
```
