```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder, OrdinalEncoder, OneHotEncoder
```

```python
df = pd.read_csv(file_path)
df
```

```python
df_copy = df.copy()
df_copy_sk = df.copy()
```

## Label Encoding (for Ordinal Data)

### Using 'map' funtion

```python
feedback_map = {
    'Poor' : 0,
    'Neutral' : 1,
    'Good' : 2,
    'Excellent' : 3
}

df_copy['Feedback Rating'] = df_copy['Feedback Rating'].map(feedback_map)

df_copy['Return Customer'] = df_copy['Return Customer'].map({'Yes': 0, 'No' : 1})
df_copy
```

### using scikit-learn

```python
feedback_order = [['Poor', 'Neutral', 'Good', 'Excellent']]
ord_encoder = OrdinalEncoder(categories=feedback_order)

df_copy_sk['Feedback Rating'] = ord_encoder.fit_transform(df_copy_sk[['Feedback Rating']])
df_copy_sk
```

```python
label_encoder = LabelEncoder()
df_copy_sk['Return Customer'] = label_encoder.fit_transform(df_copy_sk['Return Customer'])
df_copy_sk
```

```python
df_copy = pd.get_dummies(df_copy,columns=['City', 'Payment Method'], dtype=int)
df_copy
```
