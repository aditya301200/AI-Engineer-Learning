# Support Vector Regression

## Importing Libraries

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVR
```

## Importing Dataset

```python
dataset = pd.read_csv("../../Datasets/regression/SVR/Position_Salaries.csv")
X = dataset.iloc[:,1:-1].values
y = dataset.iloc[:,-1].values
```

```python
print(X)
```

```python
print(y)
```

```python
y = y.reshape(len(y),1)
```

```python
print(y)
```

## Feature Scaling

```python
sc_X = StandardScaler()
X = sc_X.fit_transform(X)

sc_y = StandardScaler()
y = sc_y.fit_transform(y)
```

```python
print(X)
```

```python
print(y)
```

## Training the SVR model on the whole dataset

```python
regressor = SVR(kernel = 'rbf')
regressor.fit(X, y)
```

## Predicting the SVR results

```python
sc_y.inverse_transform(regressor.predict(sc_X.transform([[6.5]])).reshape(-1,1))
```

## Visualising the SVR results

```python
X_plot = sc_X.inverse_transform(X)
y_plot = sc_y.inverse_transform(y)
plt.scatter(X_plot, y_plot,color='red')
plt.plot(X_plot, sc_y.inverse_transform(regressor.predict(X).reshape(-1,1)), color='blue')
plt.title("Truth or Bluff (Polynomial Regression)")
plt.xlabel("Position Level")
plt.ylabel("Salary")
plt.show()
```

## Visualising the SVR results (for higher resolution and smoother curve)

```python
X_grid = np.arange(X_plot.min(), X_plot.max(), 0.1)
X_grid = X_grid.reshape(len(X_grid), 1)
plt.scatter(X_plot, y_plot, color='red')
plt.plot(X_grid, sc_y.inverse_transform(regressor.predict(sc_X.transform(X_grid)).reshape(-1,1)), color='blue')
plt.title("Truth or Bluff (Polynomial Regression)")
plt.xlabel('Position Label')
plt.ylabel('Salary')
plt.show()
```
